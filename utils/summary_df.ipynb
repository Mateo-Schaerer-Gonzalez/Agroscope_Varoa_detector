```python
import pandas as pd

#load mite_data
mite_data = pd.read_csv('mite_data.csv')

mite_data
```

```python
# now make a summary df where for each zone you find total mites, alive mites, and dead mites,for each  recording num
summary_df = mite_data.groupby(['zone ID', 'recording']).agg(
    total_mites=('mite ID', 'count'),
    alive_mites=('status', lambda x: (x == 'alive').sum()),
    dead_mites=('status', lambda x: (x == 'dead').sum())
).reset_index()

summary_df

```
